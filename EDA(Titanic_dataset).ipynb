```python
count, bins, _ = plt.hist(df.Age, bins=20, color='blue', edgecolor='black')
bin_centers = (bins[1:] + bins[:-1]) / 2  # Calculate bin centers

plt.plot(bin_centers, count, marker='o', linestyle='-', color='red')
plt.title('Age Distribution')
plt.xticks(np.arange(0, 90, 5))

plt.show()
```

```python
vaar4 = df[df.Survived == 1].Age

count, bins, _ = plt.hist(vaar4, bins=20, color='blue', edgecolor='black')
bin_centers = (bins[1:] + bins[:-1]) / 2  # Calculate bin centers

plt.plot(bin_centers, count, marker='o', linestyle='-', color='red')
plt.title('Age Distribution')
plt.xticks(np.arange(0, 90, 5))

plt.show()
```
